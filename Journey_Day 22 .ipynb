**DATATYPES**

data types mean ki ham dibbe (variable) me kis type ka material (data) store kar rhe hai

int (Integer): Pure numbers, bina point wale.

Example: coins = 10, year = 2026

float: Point (decimal) wale numbers.

Example: price = 99.50, pi = 3.14

str (String): Koi bhi naam, text ya alphabets jo double quotes " " ya single quotes ' ' ke andar likhe hon.

Example: city = "Delhi", message = 'Hello'

bool (Boolean): Sirf do hi values—haan ya naa. Ya toh True hoga ya False.

Example: is_logged_in = True

**Implicit Typecasting**

```python
a = 10    # int
b = 2.5   # float
c = a + b # c automatic float ban jayega (12.5)
```

**Explicit Typecasting**

```python
x = "100"      # str (Text hai abhi)
num = int(x)   # Convert text "100" to Number 100
print(num + 5) # Output: 105
```

> Falsy Values in Python:
Jab hum kisi bhi non-boolean value ko bool() mein cast karte hain, toh Python in 5 things ko False manta hai, baaki sabko True:

0 (Zero integer ya float 0.0)

"" (Empty String)

None

[], (), {} (Empty Data Structures)

```python
print(bool(0))        # False
print(bool(""))       # False
print(bool("Hello"))  # True (Kyuki string khali nahi hai)
print(bool(-5))       # True (Non-zero number hai)
```

**User Input (input())**

```python
# User ne enter kiya 25
age = input("Enter your age: ") 

# age ke andar save hua "25" (text), number 25 nahi!
# print(age + 5) ❌ Error! Text mein number nahi judta.

# Sahi Tarika (Input + Typecasting)
age = int(input("Enter your age: ")) # Convert to int
print("Age after 5 years:", age + 5)
```

**input Validation & Striping**:
Real-world applications mein user typing mein extra spaces de deta hai (e.g., "  25  "). Isse typecasting fail ho sakti hai. Isliye advanced levels par hum .strip() method ka use karte hain.

```python
raw_input = "  45  "
clean_input = int(raw_input.strip()) # Spaces hata kar safe convert karega
print(clean_input) # 45
```
