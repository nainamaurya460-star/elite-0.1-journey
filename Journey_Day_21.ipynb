**Standard Output stream /sys.stdout**

print() function ka basic kaam screen par output dikhana hai. Lekin internally, iske paas kuch hidden parameters hote hain jo iske behavior ko control karte hai
$$\text{print}(\text{*objects}, \text{sep=' '}, \text{end='\textbackslash n'}, \text{file=None}, \text{flush=False})$$

 1. sep (Separator) Parameter

```python
print("data_cleaning","vibe coding","ai assisted developer")
# output: data_cleaning  vibe coding ai assisted developer
# default space ko remove karke koi bhi character add karna
print("data_cleaning","vibe coding","ai assisted developer",sep="->")
```

 2. end Parameter

```python
print("Downloading dataset....")
print("Done!!")
# output Downloading dataset...
# Done!!
# using end parameter
print("Downloading dataset...", end="")
print("Done!!") # Downloading dataset....Done!!
```

3. file Parameter (Data Logging)

```python
# Output ko screen ke bajay file mein save karna
with open("model_logs.txt", "w") as log_file:
    print("Epoch 10: Accuracy = 94.2%", file=log_file)

# Screen par kuch nahi dikhega, par 'model_logs.txt' ke andar ye text save ho jayega.
```

4. flush Parameter (Memory Buffer Bypass)

```python
import time

# Live progress bar simulator (Bina rukaawat live screen par dikhana)
print("Training Model...", end="", flush=True)
time.sleep(2) # 2 second ka gap
print(" Complete!")
```
